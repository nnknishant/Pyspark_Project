```python
data = [
    (1, 65, "Male", "Diabetes", 3000, "Hospital A"),
    (2, 72, "Female", "Hypertension", 5000, "Hospital B"),
    (3, 34, "Female", "Asthma", 1500, "Hospital A"),
    (4, 56, "Male", "Diabetes", 4000, "Hospital C"),
    (5, 29, "Female", "Flu", 200, "Hospital B"),
    (6, 51, "Male", "Hypertension", 3500, "Hospital A"),
    (7, 80, "Female", "Asthma", 2500, "Hospital C"),
    (8, 40, "Male", "Flu", 300, "Hospital B")
]
```

```python
columns = ["patient_id", "age", "gender", "diagnosis", "treatment_cost", "hospital"]
```

```python
df = spark.createDataFrame(data,columns)
```

```python
df.show()
```

```python
df1 = df
```
