```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("example").getOrCreate()
```

```python
# Import CSV File

df_pyspark = spark.read.csv('/content/test3.csv', header = True, inferSchema = True)
```

```python
df_pyspark.show()
```

```python
df= df_pyspark
```

```python
df.show()
```

```python
df.printSchema()
```

```python
# Group by

df.groupby = df.groupBy("name").sum().show()
```

```python
df_max_salary = df.groupBy("Departments").mean().show()
```

```python
df_max_salary = df.groupBy("Departments").count().show()
```

```python
# Group by by department which gives maximum  salar
df_max_salary = df.groupBy("Departments").sum().show()
```

```python
df_total_Salary =  df.agg({"salary": "sum"}).show()
```

```python
df_max_salary = df.groupBy("name").max().show()
```

```python
df_avg_salary = df.groupBy("Name").avg().show()
```
