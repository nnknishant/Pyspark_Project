```python
# Creating spark session

from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("example").getOrCreate()
```

```python
# Reading the data
df = spark.read.csv('/kaggle/input/getdata', header = True, inferSchema = True)
```

```python
# Looking the first row of the dataset

df1 = df
```

```python
# To see the entire columns

df1.show(2)
```

```python
# shape of the dataset
print((df1.count(), len(df1.columns)))
```

```python
df1.take(2)
```

```python
df1.describe().show(truncate = True)
```

```python
# to select the specific columns

df1.select("Pregnancies", "Age").show(10)
```

```python
# To select the col method here

from pyspark.sql.functions import col
```

```python
df.select(col("Pregnancies"), col("Age")).show(2)
```

```python
# now  data filtering
df1.filter(df1["Age"] > 30).show()
```

```python
df1.filter(df1["Age"] <= 50).show(10)
```

```python
# Now filter the data of two columns
df1.filter(df1['Age'] < 30).select("Insulin", "BMI", "Outcome").show()
```

```python
# filter on multiple conditions
df1.filter(df1['Age'] > 60).filter(df1['Outcome'] == '1').show()
```

```python
# Conting after filter
df1.where(df1['Outcome'] == 1).show()
```

```python
df1.where((df1['Outcome'] ==1) & (df1['Pregnancies'] >= 9)).show(10)
```

```python
df1.where(df1['Age'] > 50).count()
```

```python
# Now add the columns
df1.withColumn("New Age", df1["Age"] + 10).show()
```

```python
df1.withColumn("Updated BMI", df1["BMI"] + 10).show()
```

```python
# Grouping the dataset
df1.groupby("Outcome").count().show()
```

```python
df1.groupby ("Age").count().show()
```

```python
# To check the distinct values
df1.select("Pregnancies").distinct().count()
```

```python
df1.select("Pregnancies", "Age").distinct().count()
```

```python
df1.select(col("Pregnancies"), col("Age")).distinct().count()
```

```python
# Check the dataset by aggregating functions
import pyspark.sql.functions as F

df1.groupby("Outcome").agg(F.mean("Age").alias("Age_mean")).show()
```

```python
# Now  apply here user defined functions

def diabetes(case):
    if case == 1:
        return "disbetes"
    else:
        return "no disbetes"

```

```python
from pyspark.sql.types import*
diabetes_udf = F.udf(diabetes, StringType())
df1.withColumn("Diabetes_case", diabetes_udf(df1["Outcome"])).show(10)
```
