```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("demo").getOrCreate()
```

```python
data  = [("Alice", 22), ("Nishant", 23), ("Rupam", 25), ("Anjali",29), ("Preeti", 21)]

columns = ["Name", "Age"]

df = spark.createDataFrame(data, columns)
```

```python
df.show()
```

```python
#Using withColumn to Add or Modify a Column
from pyspark.sql.functions import col

df = df.withColumn("Age_10_years_later",col("Age")+10)
```

```python
df.show()
```

```python
# Filtering Rows

df.filter(col("Age")> 25).show()
```

```python
# Group By and Aggregate

data = [("HR", 3000), ("IT", 4000), ("HR", 3500), ("IT", 4500)]
df = spark.createDataFrame(data, ["Department", "Salary"])
```

```python
df.show()
```

```python
df.grouped_by = df.groupBy("Department").avg("Salary").show()
```

```python
 # Using when and otherwise for Conditional Logic

from pyspark.sql.functions import  when
```

```python
df = df.withColumn("seniority", when(col("Salary") > 4000, "ssenior").otherwise("junior"))
df.show()
```

```python
# Joins the Dataframes

employees = [("Alice", "HR"), ("Bob", "IT"), ("Cathy", "HR")]
salaries = [("Alice", 3000), ("Bob", 4000), ("Cathy", 3500)]

df_emp = spark.createDataFrame(employees, ["Name", "Department"])
df_sal = spark.createDataFrame(salaries, ["Name", "Salary"])
```

```python
df_emp.show()
```

```python
df_sal.show()
```

```python
df_joined = df_emp.join(df_sal, on="Name", how = "inner").show()
```

```python
df_joined = df_emp.join(df_sal, on="Name", how = "left").show()
```
